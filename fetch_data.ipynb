```python
from pybaseball import batting_stats
import pandas as pd
```

```python
selected_cols = ['1','2','3','4','5','6','7', '11', '16', '21', '39', '40', '50', '58', '60', '61', 'DEFENSE', 'FIELDING']
```

```python
train_data = batting_stats(2000, 2009, qual = 100, stat_columns = selected_cols)
```

```python
train_data.shape
```

```python
train_data2 = batting_stats(2010, 2019, qual = 100, stat_columns = selected_cols)
```

```python
train_data2.shape
```

```python
train_data3 = batting_stats(1980, 1989, qual = 100, stat_columns = selected_cols)
```

```python
train_data3.shape
```

```python
train_data4 = batting_stats(1990, 1999, qual = 100, stat_columns = selected_cols)
```

```python
all_train_data = pd.concat([train_data3, train_data4, train_data, train_data2], axis = 0).reset_index(drop = True)
all_train_data.shape
```

```python
all_train_data.to_csv('hitter_train_data.csv', index = False)
```
